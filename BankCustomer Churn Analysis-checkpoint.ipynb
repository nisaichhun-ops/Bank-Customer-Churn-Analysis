```python
df.groupby('Geography').agg({
    'Exited': 'mean',
    'Age': 'mean',
    'Balance': 'mean',
    'CreditScore': 'mean',
    'EstimatedSalary': 'mean',
    'IsActiveMember': 'mean'
}).round(2)
```

```python
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[18, 30, 40, 50, 60, 100],
    labels=['18-30', '31-40', '41-50', '51-60', '60+'],
    include_lowest=True
)
```

```python
df.groupby('AgeGroup', observed=True).agg(
    Customers=('Exited', 'count'),
    ChurnRate=('Exited', 'mean')
).assign(
    ChurnRate=lambda x: (x['ChurnRate'] * 100).round(2)
)
```

```python
df['BalanceGroup'] = pd.cut(
    df['Balance'],
    bins=[-1, 0, 50000, 100000, 150000, 250000],
    labels=['Zero', 'Low', 'Medium', 'High', 'Very High']
)

df.groupby('BalanceGroup', observed=True).agg(
    Customers=('Exited', 'count'),
    ChurnRate=('Exited', 'mean')
).assign(
    ChurnRate=lambda x: (x['ChurnRate'] * 100).round(2)
)
```
